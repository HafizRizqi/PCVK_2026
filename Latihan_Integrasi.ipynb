```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt


```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = [
    "https://picsum.photos/200/300.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
    "https://picsum.photos/600/400.jpg"
]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                      # read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5)  # resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   # convert color to RGB
    final_frame = cv.hconcat([image, image_2])       # concatenate image
    cv2_imshow(final_frame)                          # show image
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal di tengah image
for y in range(lebar):
    image_3[int((tinggi)/2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
# Membaca gambar
image = io.imread("https://picsum.photos/200/300.jpg")

# Salin gambar
hasil = image.copy()

# Mendapatkan ukuran gambar
height, width = hasil.shape[:2]

# Warna putih
white = [255, 255, 255]
```

```python
# Garis vertikal di tengah
x_tengah = width // 2

for y in range(height):
    hasil[y, x_tengah] = white
```

```python

# Gabungkan gambar asli dan hasil
gabungan = np.hstack((image, hasil))

cv2_imshow(gabungan)
```

```python
# Salin gambar
hasil2 = image.copy()

# Diagonal kiri atas - kanan bawah
for i in range(min(height, width)):
    hasil2[i, i] = white

# Diagonal kanan atas - kiri bawah
for i in range(min(height, width)):
    hasil2[i, width - 1 - i] = white
```

```python
# Gabungkan gambar asli dan hasil
gabungan = np.hstack((image, hasil2))

cv2_imshow(gabungan)
```

```python
hasil3 = image.copy()

height, width = hasil3.shape[:2]

y_tengah = height // 2
panjang = 150

x_awal = (width - panjang) // 2
x_akhir = x_awal + panjang

for x in range(x_awal, x_akhir):
    hasil3[y_tengah, x] = [255, 255, 255]

gabungan = np.hstack((image, hasil3))

cv2_imshow(gabungan)
```

```python
hasil4 = image.copy()

# Mendapatkan ukuran gambar
tinggi_gambar, lebar_gambar = hasil4.shape[:2]

# Ukuran kotak
lebar = 80
tinggi = 80

# Posisi kotak di tengah gambar
x1 = (lebar_gambar - lebar) // 2
y1 = (tinggi_gambar - tinggi) // 2

# Membuat kotak putih
for y in range(y1, y1 + tinggi):
    for x in range(x1, x1 + lebar):
        hasil4[y, x] = [255, 255, 255]

# Menggabungkan gambar asli dan hasil
gabungan = np.hstack((image, hasil4))

cv2_imshow(gabungan)
```
