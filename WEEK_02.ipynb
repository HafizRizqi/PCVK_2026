```python
# Membaca gambar
img = cv2.imread('/content/drive/MyDrive/KTM lama.jpg')

# Konversi BGR ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ==========================
# MERAH + BIRU
# ==========================
rb = img_rgb.copy()

# Kanal hijau dibuat 0
rb[:, :, 1] = 0

# ==========================
# HIJAU + BIRU
# ==========================
gb = img_rgb.copy()

# Kanal merah dibuat 0
gb[:, :, 0] = 0

# ==========================
# TAMPILKAN BERDAMPINGAN
# ==========================
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(rb)
plt.title("Kombinasi Kanal Merah-Biru")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gb)
plt.title("Kombinasi Kanal Hijau-Biru")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Membaca gambar
img = cv2.imread('/content/drive/MyDrive/KTM lama.jpg')

# Konversi agar warna benar saat ditampilkan
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Pencerminan vertikal
flip_vertical = cv2.flip(img_rgb, 0)

# Pencerminan horizontal
flip_horizontal = cv2.flip(img_rgb, 1)

# Pencerminan vertikal + horizontal
flip_both = cv2.flip(img_rgb, -1)

# Menampilkan dalam satu figure
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertical)
plt.title("Pencerminan Vertikal")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Pencerminan Horizontal")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title("Pencerminan Vertikal + Horizontal")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
def tambah_label(img, nama):

    # Mendapatkan ukuran gambar
    tinggi, lebar = img.shape[:2]

    # ==========================
    # Ukuran bar berdasarkan gambar
    # ==========================

    tinggi_bar = int(tinggi * 0.12)

    # Posisi awal bar
    y_bar = tinggi - tinggi_bar

    # Membuat bar hitam menggunakan slicing
    img[y_bar:tinggi, :] = 0

    # ==========================
    # Ukuran font berdasarkan tinggi gambar
    # ==========================

    font_scale = tinggi / 800

    # Ketebalan teks berdasarkan tinggi gambar
    thickness = max(1, int(tinggi / 400))

    # Posisi teks berdasarkan ukuran gambar
    x_text = int(lebar * 0.03)
    y_text = int(tinggi - tinggi_bar * 0.30)

    # ==========================
    # Menambahkan nama
    # ==========================

    cv2.putText(
        img,
        nama,
        (x_text, y_text),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (255, 255, 255),
        thickness
    )

    return img


# ==================================
# Contoh gambar pertama
# ==================================

img1 = cv2.imread('/content/drive/MyDrive/KTM lama.jpg')

hasil1 = tambah_label(
    img1.copy(),
    "Joko"
)


# ==================================
# Contoh gambar kedua
# ==================================

img2 = cv2.imread('/content/drive/MyDrive/IFriend.jpg')

hasil2 = tambah_label(
    img2.copy(),
    "Hafiz Rizqi Hernanda"
)


# ==================================
# Menampilkan hasil
# ==================================

from google.colab.patches import cv2_imshow

print("Gambar 1:")
cv2_imshow(hasil1)

print("Gambar 2:")
cv2_imshow(hasil2)
```
